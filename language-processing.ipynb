```python
import nltk
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
sentence = "Hi, my name is Mohammad Reza. I'm learning python to improve myself."
```

```python
tokens = nltk.word_tokenize(sentence)
print(tokens)
```

```python
'''
NN = Noun
VBZ = Verb, third-person singular
VBG = Verb ending in -ing
JJ = Adjective
NNP = Proper noun
'''

pos_tags = nltk.pos_tag(tokens)
print(pos_tags)
```
